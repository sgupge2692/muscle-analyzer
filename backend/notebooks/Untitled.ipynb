```python
import pickle, os, re

skl_dir = '/Users/ryota/Desktop/muscle-analyzer/data/raw/musclemap_benchmark/musclemap136_skl/'
ann_base = '/Users/ryota/Desktop/muscle-analyzer/data/raw/musclemap_benchmark/annotation/'

# trainのfile_listとlabel_listを読む
with open(ann_base + 'MuscleMap136_train_file_list.pkl', 'rb') as f:
    train_files = pickle.load(f)
with open(ann_base + 'MuscleMap136_train_label_list.pkl', 'rb') as f:
    train_labels = pickle.load(f)

# クラスID→種目名のマッピングを作成
class_to_exercise = {}
for fname, label in zip(train_files, train_labels):
    skl_name = fname.replace('.mp4', '.pkl')
    skl_path = skl_dir + skl_name
    if os.path.exists(skl_path) and label not in class_to_exercise:
        try:
            with open(skl_path, 'rb') as f:
                skl_data = pickle.load(f)
            match = re.search(r'AN_(.+?)_ST_', skl_data['frame_dir'])
            if match:
                class_to_exercise[label] = match.group(1)
        except:
            pass

print(f"マッピング数: {len(class_to_exercise)}")
for cid, name in sorted(class_to_exercise.items()):
    print(f"  クラス{cid:4d}: {name}")
```

```python
import json, pickle, os, re

# マッピング読み込み
with open('/Users/ryota/Desktop/muscle-analyzer/data/raw/muscle_mapping.json', 'rb') as f:
    mapping = json.load(f)

exercise_to_muscles = mapping['exercise_muscle_mapping']

# sklから種目名を取得して筋群ラベルに変換
skl_dir = '/Users/ryota/Desktop/muscle-analyzer/data/raw/musclemap_benchmark/musclemap136_skl/'

results = []
for fname in sorted(os.listdir(skl_dir)):
    try:
        with open(skl_dir + fname, 'rb') as f:
            data = pickle.load(f)
        match = re.search(r'AN_(.+?)_ST_', data['frame_dir'])
        if match:
            exercise = match.group(1)
            muscles = exercise_to_muscles.get(exercise, [])
            results.append({
                'file': fname.replace('.pkl', '.mp4'),
                'exercise': exercise,
                'muscle_labels': muscles
            })
    except:
        pass

print(f"総サンプル数: {len(results)}")
print(f"\n例:")
for r in results[:3]:
    print(f"  {r['file']}")
    print(f"  種目: {r['exercise']}")
    print(f"  筋群: {r['muscle_labels']}")
```

```python
import pickle

# 保存
output_path = '/Users/ryota/Desktop/muscle-analyzer/data/processed/musclemap_labeled.pkl'
with open(output_path, 'wb') as f:
    pickle.dump(results, f)

print(f"保存完了: {len(results)}サンプル")
```
